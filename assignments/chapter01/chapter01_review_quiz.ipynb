2. 나이를 이름과 함께 나란히 나열한다.

4. 첫번째 사람을 일단 최대, 최소 종이에 이름과 나이를 둘 다 적어놓는다.
5. 다음 사람이 들어오면 최대값에 있는 종이 나이와 최소값에 있는 종이와 비교하며 나이가 더 많다면 최대값으로 이동하여 기존 값을 지우고, 나이가 더 적다면 최소값으로 이동하여 기존값을 지운다.

튜터님 답:
1. 회원들의 이름과 나이 목록을(list) 준비한다.
2. 종이 2장을 준비한다.(종이에 내용은 없다. 최대 최소 나이 적어야 함.) ==>0
3. 회원  한 명씩 순회하며 아래 내용을 진행한다.
	3.1현재 회원의 나이를 최대값 종이의 값과 비교
		3.2 만약 최대값보다 크면 최대값 종이의 값을 바꿈.
		3.3 만약 최소값보다 작으면 최소값 종이의 값을 바꿈.
4. 종이 내용을 보고

```python
#records
# 2. 종이 2장을 준비한다.(종이에 내용은 없다. 최대 최소 나이 적어야 함.) ==>0
max_age = 0
min_age = 1000
# 3. 회원  한 명씩 순회하며 아래 내용을 진행한다.
for rec in records:
	age = rec["Age"]
	# 3.1현재 회원의 나이를 최대값 종이의 값과 비교
	if age > max_age:
		# 3.2 만약 최대값보다 크면 최대값 종이의 값을 바꿈.
		max_age = age
		# 3.3 만약 최소값보다 작으면 최소값 종이의 값을 바꿈.
	if age < min_age:
		min_age = age
#4. 종이 내용을 보고
print("max_age : ", max_age)
print("min_age : ", min_age)
```
