### 가장 어려웠던 문제와 이유
영어실력이 부족하여 변수정하기와 영어타자치기가 가장 어려웠습니다.
그리고 코드를 짜면서 :이나 ,을 어디에 어떻게 쓰는지 잘 숙지하지 못하여 오류가 나는 경우가 종종 있어서 어려웠습니다.

```python
# Chapter 02 준비 코드 — 반드시 직접 입력

students = [
    {"name":"민수","score":95,"attendance":90},
    {"name":"지영","score":85,"attendance":95},
    {"name":"현우","score":72,"attendance":65},
    {"name":"수진","score":78,"attendance":88}
]
for student in students:
    name = student["name"]
    score = student["score"]
    attendance = student["attendance"]
    if score < 60 or attendance < 70:
        grade = "F"
    elif score >= 90 and attendance >=80:
        grade = "A"
    elif score >= 80 and attendance >=80:
        grade = "B"
    elif score >= 70:
        grade = "C"
    else:
        grade="D"
    print(
        f"{name}: 점수{score}점"
        f"출석률 {attendance}%, 등급{grade}"
    )
```
