```python
# 4-1. 점수 판단
students = [
    {"name":"민수", "score":85},
    {"name":"지영", "score":80},
    {"name":"현우", "score":79}
]
for student in students:
    name = student["name"]
    score = student["score"]

    if score >= 80:
        result = "합격"
    else:
        result = "불합격"
    print(name, score, result)
```

```python
# 4-2. 주문 금액 판단
orders = [
    {"customer":"민수", "amount":49000},
    {"customer":"지영", "amount":50000},
    {"customer":"현우", "amount":51000}
]
for order in orders:
    customer = order["customer"]
    amount = order["amount"]
    if amount >= 50000:
        result = "무료배송"
    else:
        result = "배송비 있음"
    print(customer, amount, result)
```

### 경계값 확인

- 50,000원 결과:

- 51,000원 결과:

### 2. 전체 구조

- 데이터:학생의 이름과 점수
(name=민수, 지영, 현우, 수진
score=85, 72, 91, 80)

- 반복:학생을 한 명씩 확인

### 3. 처리 순서

1.학생 이름과 점수를 순서대로 나열한다.

2.한 명씩 점수를 확인한다. 

3.80점 이상이면 합격, 미만이면 불합격인 조건을 확인한다.

### 5. 코드 작성 전 확인

- List가 필요한 이유:여러 Dictionary를 하나로 묶어서 순서대로 반복하기 위해서다.

- Dictionary가 필요한 이유:한 명의 이름과 점수처럼 여러 데이터 정보를 key와 value로 구분해서 저장하기 위해서다.

- for가 필요한 이유:데이터를 하나씩 꺼내오기 위해서다.

- if/else가 필요한 이유:조건 해당 유무를 확인하고 나누기 위해서다.

- 최종 출력에 필요한 값:이름, 점수, 결과


```python
students = [
    {"name":"민수", "score":85},
    {"name":"지영", "score":72},
    {"name":"현우", "score":91},
    {"name":"수진", "score":80}
]
for student in students:
    name = student["name"]
    score = student["score"]
    if score >= 80:
        result = "합격"
    else:
        result = "불합격"
    print(name, score, result)
```
