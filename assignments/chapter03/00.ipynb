```python
# 6. 학생 성적을 세 단계로 분류하기
# 요구사항
# 다음 다섯 학생의 이름과 점수가 주어져 있습니다.

students = [
    {"name":"민수","score":79},
    {"name":"지영","score":80},
    {"name":"현우","score":89},
    {"name":"수빈","score":90},
    {"name":"원빈","score":91}
]
print(students)
print(type(students))

 

# 모든 학생을 한 명씩 확인하여 다음 기준으로 결과를 정합니다.
for student in students:
    # print(student)

    if student["score"] >= 90:
        print(student["name"],student["score"],"우수")
# • 90점 이상이면 우수
    elif student["score"] >= 80:
        print(student["name"],student["score"],"합격")
# • 80점 이상이면 합격
    else:
        print(student["name"],student["score"],"보완 필요")
# • 그 외에는 보완 필요

# • 모든 학생의 이름, 점수, 결과를 확인할 수 있어야 합니다.
```

```python
students = [
    {"name":"민수","score":79},
    {"name":"지영","score":80},
    {"name":"현우","score":89},
    {"name":"수빈","score":90},
    {"name":"원빈","score":91}
]
for student in students:
    name = student["name"]
    score = student["score"]
    if score >= 90:
        result = "우수"
    elif score >= 80:
        result = "합격"
    else:
        result = "보완필요"
    print(name, score, result)
    
```

```python
customers = [
    {"name":"민수","amount":49000,"grade":"일반"},
    {"name":"지영","amount":50000,"grade":"VIP"},
    {"name":"현우","amount":99000,"grade":"일반"},
    {"name":"수빈","amount":100000,"grade":"VIP"},
    {"name":"원빈","amount":120000,"grade":"일반"}
]
for customer in customers:
    name = customer["name"]
    amount = customer["amount"]
    grade = customer["grade"]
    ship_type = ""
    if amount >= 100000 and grade == "VIP":
        ship_type  = "우선배송"
    elif amount >= 50000 and grade == "VIP":
        ship_type = "무료배송"
    else:
        ship_type = "일반배송"
    print(f"{name}님의 배송 결과는 {ship_type} 입니다. 주문금액 : {amount}원, 회원등급 : {grade}")
```

```python
products = [
    {"name":"키보드", "grade":4.8, "stock":12, "state":"판매중"},
    {"name":"마우스", "grade":4.6, "stock":3, "state":"판매중"},
    {"name":"모니터", "grade":4.2, "stock":0, "state":"판매중"},
    {"name":"헤드셋", "grade":4.9, "stock":8, "state":"판매중지"},
    {"name":"웹캠", "grade":3.8, "stock":10, "state":"판매중"}
]
for product in products:
    name = product["name"]
    grade = product["grade"]
    stock = product["stock"]
    state = product["state"]
    if state != "판매중" or stock <= 0:
        result = "판매불가"
    elif grade >= 4.5 and stock >= 5:
        result = "추천"
    elif grade >= 4.0:
        result = "일반"
    else:
        result = "관심 필요"
    print(f"{name} 평점은 {grade}이며, 재고는 {stock}입니다. 현재상태는 {state}이며, {result} 합니다.")
    
```
